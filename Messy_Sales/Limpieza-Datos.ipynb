1. Cargar librerias

```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

2. Cargar dataset

```python
ruta ='/content/drive/MyDrive/U/Fundamentos de Ciencia de Datos/Messy_Sales/messy_sales.xlsx'
df = pd.read_excel(ruta)
df.head()
```

3. Revisar información

```python
df.info()
```

```python
df.isnull().sum()
```

4. Gestión de datos faltantes

```python
## Asegurar que las columnas numericas si sean numeros
df['Units Sold'] = pd.to_numeric(df['Units Sold'], errors='coerce')
df['Total Sales'] = pd.to_numeric(df['Total Sales'], errors='coerce')
```

```python
df.head()
```

5. Llenar valor numericos con el promedio

```python
df['Units Sold'] = df['Units Sold'].fillna(df['Units Sold'].mean())
df['Total Sales'] = df['Total Sales'].fillna(df['Total Sales'].mean())
```

6.

```python
df.columns = df.columns.str.strip()
df.head()
```

7. Arreglar columna date

```python
df.Date = pd.to_datetime(df.Date, errors='cource')
df.head()
```

Estandarizar el texto de las columnas

```python
for col in ["Region","Product","Category"]:
  df[col] = df[col].str.strip().str.title()
```

```python
df.head()
```

8. Limpiar columna currency
(Unit Price, Total Sales)

```python
for col in ["Unit Price", "Total Sales"]:
  df[col] = df[col].astype(str)
  df[col] = (
      df[col]
      .str.replace("$", "", regex=True)
      .str.replace("USD", "", regex=True)
      .str.strip()
      )
  df[col] = pd.to_numeric(df[col], errors="coerce")
df.head()
```

9. Arreglar Profit Margin


```python
df['Profit Margin (%)'] =(
    df['Profit Margin (%)']
    .astype(str)
    .str.replace("%", "", regex=True)
    .str.replace("percent","",regex=True)
    .str.strip()
    .astype(float)
)
#Si es menor a 1 multiplicar opr 100 sino no hacer nada
df['Profit Margin (%)'] = np.where(df['Profit Margin (%)']<1, df['Profit Margin (%)']*100, df['Profit Margin (%)'])
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

10. Graficar y Analizar

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df,x="Region",y="Total Sales", estimator=sum, palette="cool")
plt.title("Ventas totales por region", fontsize = 14)
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,6))
df.groupby("Date")["Total Sales"].sum().plot(kind="line", color="teal")
plt.title("Tendencia de ventas en el tiempo", fontsize=14)
plt.ylabel("Ventas totales (USD)", fontsize=14)
plt.xticks(rotation=30)
plt.show()
```

Tarea 1:
Realizar analisis por region

```python
plt.figure(figsize=(10,6))

ventas_region = df.groupby(["Date", "Region"])["Total Sales"].sum().unstack()
ventas_region.plot(kind="line", ax=plt.gca())

plt.title("Tendencia de ventas por región", fontsize=14)
plt.ylabel("Ventas totales (USD)", fontsize=14)
plt.xticks(rotation=30)
plt.legend(title="Región")
plt.show()
```

Analisis de torta

```python
plt.figure(figsize=(6,6))
category_sales=df.groupby("Category")["Total Sales"].sum()
category_sales.plot(kind="pie",autopct="%1.1f%%",startangle=90,color=sns.color_palette("pastel"))
plt.title("Ventas por Categoría")
plt.ylabel("")
plt.show()
```

```python
df['Category'].replace('Gadgets', 'Gadget', inplace=True)
df['Category'].replace('Elec', 'Electronics', inplace=True)
df['Category'].unique()
```

```python
plt.figure(figsize=(6,6))
category_sales=df.groupby("Category")["Total Sales"].sum()
category_sales.plot(kind="pie",autopct="%1.1f%%",startangle=90,color=sns.color_palette("pastel"))
plt.title("Ventas por Categoría")
plt.ylabel("")
plt.show()
```
