```python
from collections import Counter
import re
```

```python
with open("/content/drive/MyDrive/Colab Notebooks/Fundamentos de Ciencia de Datos/Clase-7-sept/elFracasoIA.txt", "r", encoding="utf-8") as archivo:
  discurso = archivo.read()
```

```python
len(discurso)
```

```python
discurso = discurso.lower()
discurso = discurso.replace(",", "")
discurso = discurso.replace(";", "")
discurso = discurso.replace(".", "")
discurso = discurso.replace(":", "")
discurso = discurso.replace("-", "")
discurso = discurso.replace("—", "")
discurso = discurso.replace("¿", "")
len(discurso)
```

```python
# Normalizar: minúsculas y extraer solo palabras (sin puntuación)
palabras = re.findall(r'\b[a-záéíóúñü]+\b', discurso.lower())

# Contar frecuencias
frecuencias = Counter(palabras)

# Mostrar las 10 palabras más comunes
for palabra, cantidad in frecuencias.most_common(10):
    print(f"{palabra}: {cantidad}")
```
