## Reglas para que el análisis sea correcto

**Qué tabla usar**
- Variables de **persona** (edad, sexo, educación, salud, trabajo, ingresos): `df_personas`.
- Variables del **hogar** (tamaño, gastos, vivienda, bienes, eventos, clasificación Sisbén): `df_hogares`, que tiene una fila por hogar.

**Etiquetas de los códigos**
- Siguen la ficha del **Sisbén IV**. Antes de usar una etiqueta, se verifica con los datos que sea coherente (7.1.1, 7.2.1, 7.3.1, 7.4.1 y 7.5.1).

**Nulos "por diseño"** (6.3 y 6.4)
- Las 42 variables de `NO_APLICA` tienen nulos donde la pregunta no aplicaba. Sus porcentajes se calculan **sobre quienes respondieron**: `distribucion()` ya descarta los nulos. Al reportar, decir sobre cuántas personas se calcula (por ejemplo, el analfabetismo sobre personas de 15 años o más).
- En `grado_alcanzado` el 9 es el grado noveno y en `tip_empleado` es una categoría válida.

**Comparación entre veredas**
- Comparar con **porcentajes y medianas**, nunca con conteos: Laguna Baja tiene 517 personas y Laguna Alta 119.
- En dinero usar **medianas**: hay valores extremos sospechosos que no se pudieron corregir (6.4). Los **ingresos están subdeclarados**, así que la capacidad económica se compara con el **gasto** (7.3.4).
- **La Laguna:** analizar Alta y Baja por separado y también juntas.
- En Laguna Alta (52 hogares) cada hogar pesa casi 2 puntos porcentuales: solo las diferencias grandes son concluyentes.
- Indicadores con **pocos casos** (discapacidad: 47; eventos: 50 hogares): las diferencias pequeñas entre veredas no son concluyentes.

#### Eliminación de columnas (paso 6.1)
Desde aquí se trabaja sobre **`df_limpio`**, una copia de `df_veredas`. Cada paso de la fase 2 elimina las columnas que ya sabemos que no sirven.

La función `eliminar()` quita las columnas y anota el **motivo** en el registro `eliminadas`. Así, al final se puede consultar qué se quitó y por qué.

```python
eliminadas = {}  # registro: columna -> motivo por el que se eliminó

def eliminar(df, columnas, motivo):
    """Quita las columnas de df y anota el motivo en el registro `eliminadas`."""
    columnas = [c for c in columnas if c in df.columns]
    eliminadas.update(dict.fromkeys(columnas, motivo))
    print(f"- {len(columnas)} columnas eliminadas ({motivo}). Quedan {df.shape[1] - len(columnas)}.")
    return df.drop(columns=columnas)

df_limpio = df_veredas.copy()

df_limpio = eliminar(df_limpio, calidad.index[calidad['estado'] == 'vacía'], 'vacía')
df_limpio = eliminar(df_limpio, calidad.index[calidad['estado'] == 'constante'], 'constante')
df_limpio = eliminar(df_limpio, calidad.index[calidad['estado'] == 'con nulos'],
                     'con nulos: datos personales o fechas del operativo')

print(f"\ndf_limpio: {df_limpio.shape[0]} filas x {df_limpio.shape[1]} columnas")
```

### 6.3 Códigos "no aplica"
En el censo, cuando una pregunta no le corresponde a una persona u hogar (por ejemplo, preguntarle a un niño si trabaja), no se deja vacía: se llena con un **código**, normalmente **9** o **99**. Para pandas ese 9 es un número como cualquier otro, así que entraría en promedios y porcentajes como si fuera una respuesta real.

**Cómo se hace:**
1. **Detectar candidatas:** columnas de códigos (pocos valores distintos) cuyo valor **más alto** es 9 o 99. En una pregunta de sí/no (1 = sí, 2 = no) o de opciones (1 a 8), un 9 o 99 al final es la señal típica de "no aplica".
2. **Revisar cada una:** ver su distribución y decidir. No todo 9 es "no aplica": en una variable con opciones de 1 a 10, el 9 puede ser una opción real.
3. **Reemplazar por nulo** solo donde se confirme.

```python
# 1. Candidatas: columnas numéricas con pocos valores distintos (códigos) cuyo valor máximo es 9 o 99
numericas = df_limpio.select_dtypes('number')
codigos = numericas.columns[numericas.nunique() <= 20]          # columnas de códigos, no de cantidades
candidatas = [c for c in codigos if df_limpio[c].max() in (9, 99)]

print(f"Columnas de códigos: {len(codigos)} | Candidatas con 9 o 99 como valor máximo: {len(candidatas)}\n")

# 2. Distribución de cada candidata: {código: número de filas}
for c in candidatas:
    print(f"{c:26s} {df_limpio[c].value_counts().sort_index().to_dict()}")
```

Salen **44 candidatas**. Para decidir con evidencia, en las variables de **persona** se mira la **edad** de quienes tienen el código. Si el código "no aplica" está bien usado, debería coincidir con grupos lógicos: por ejemplo, niños pequeños en las preguntas de educación o hombres en la de embarazo.

```python
# 3. Evidencia: ¿a quién le aparece el código? (solo variables de persona)
filas = []
for c in candidatas:
    if not varia_en_hogar.get(c, False):      # se omiten las variables del hogar
        continue
    for codigo in (9, 99):
        tiene = df_limpio[c] == codigo
        if tiene.any():
            edades = df_limpio.loc[tiene, 'edad_calculada']
            filas.append({'variable': c, 'código': codigo, 'filas': tiene.sum(),
                          'edad mín.': edades.min(), 'edad máx.': edades.max(),
                          '% hombres': round((df_limpio.loc[tiene, 'sexo_persona'] == 1).mean() * 100)})

pd.DataFrame(filas)
```

### 6.4 Valores atípicos
Un valor atípico es uno muy alejado del resto, por ejemplo un gasto de $6.000.000 cuando la mayoría gasta $300.000. Puede ser un **error de digitación** o un **caso real pero excepcional**. En ambos casos distorsiona los promedios, por eso hay que encontrarlos antes de analizar.

**Cómo se detectan: regla del rango intercuartílico (IQR)**
- Q1 y Q3 son los valores que dejan el 25 % y el 75 % de los datos por debajo. IQR = Q3 − Q1 es el ancho del 50 % central.
- Un valor es **extremo** si está por encima de **Q3 + 3 × IQR** o por debajo de **Q1 − 3 × IQR**. Se usa 3 en lugar del 1,5 habitual para marcar solo los casos realmente alejados.

**Dos cuidados:**
- Las variables del **hogar** se revisan con **una fila por hogar**. Si no, un hogar de 7 personas contaría 7 veces.
- En **dinero**, la mayoría de las filas tiene 0 (no tiene ese gasto o ese ingreso). La regla se aplica solo a los **valores mayores que 0**. Si no, cualquier valor positivo saldría como atípico.

```python
# 1. Variables de cantidad (no de códigos): edad, número de cosas (num_) y valores en pesos (vlr_)
cantidades = [c for c in df_limpio.columns
              if c == 'edad_calculada' or c.startswith(('num_', 'vlr_'))]
cantidades.remove('num_ficha')  # es un identificador, no una cantidad

# 2. Tabla con una fila por hogar, para revisar las variables del hogar
hogares = df_limpio.drop_duplicates('id_hogar')

def revisar(c):
    """Resumen de una variable y cuántos valores extremos tiene según la regla 3 × IQR."""
    del_hogar = not varia_en_hogar[c]
    datos = (hogares if del_hogar else df_limpio)[c].dropna()
    if c.startswith('vlr_'):
        datos = datos[datos > 0]                      # en dinero, solo valores positivos
    q1, q3 = datos.quantile([0.25, 0.75])
    iqr = q3 - q1
    extremos = datos[(datos > q3 + 3 * iqr) | (datos < q1 - 3 * iqr)]
    return {'unidad': 'hogar' if del_hogar else 'persona', 'datos': len(datos),
            'mín.': datos.min(), 'mediana': datos.median(), 'máx.': datos.max(),
            'límite superior': q3 + 3 * iqr, 'extremos': len(extremos)}

atipicos = pd.DataFrame({c: revisar(c) for c in cantidades}).T
with pd.option_context('display.float_format', '{:,.0f}'.format):
    display(atipicos.sort_values('extremos', ascending=False))
```

**Lectura de la tabla:** no todos los "extremos" son problemas reales. Hay que separar cuatro situaciones:

1. **Un código "no aplica" que se escapó:** `num_sem_buscando` tiene **mediana 999**. El 999 tiene tres dígitos, así que el paso 6.3 (que buscaba 9 y 99) no lo detectó.
2. **Columnas que no son cantidades:** `num_tel_contacto` (teléfono, dato personal) y `num_paquete` (control del operativo) empiezan por `num_`, pero no miden nada.
3. **Falsas alarmas en conteos pequeños:** en `num_cuartos_dormir` y `num_evento_*` casi todos tienen el mismo valor (1 cuarto, 0 eventos), así que Q1 = Q3 y el IQR es 0. Con un IQR de 0, cualquier valor distinto sale "extremo". Tener 2 cuartos o haber sufrido un vendaval **no es un error**.
4. **Dinero:** gastos e ingresos muy altos. Hay que revisarlos uno por uno.

La siguiente celda revisa los puntos 1, 2 y 4, y comprueba además que el gasto total coincida con la suma de los gastos.

```python
# 1. num_sem_buscando: ¿qué valores tiene además del 999?
print("num_sem_buscando:", df_limpio['num_sem_buscando'].value_counts().sort_index().to_dict())

# 2. ¿num_personas_posibles y num_habita_vivienda repiten el tamaño del hogar?
for c in ['num_personas_posibles', 'num_habita_vivienda']:
    print(f"{c} = num_personas_hogar en {(hogares[c] == hogares['num_personas_hogar']).mean():.0%} de los hogares")

# 3. ¿El gasto total es la suma de los gastos?
gastos = [c for c in hogares.columns if c.startswith('vlr_gasto_')]
diferencia = hogares['vlr_total_gastos'] - hogares[gastos].sum(axis=1)
print(f"\nHogares donde vlr_total_gastos ≠ suma de los gastos: {(diferencia != 0).sum()} de {len(hogares)}")

# 4. Hogares con algún gasto por encima de su límite superior
limites = atipicos['límite superior']
dinero = gastos + ['vlr_total_gastos']
con_extremo = (hogares[dinero] > limites[dinero]).any(axis=1)
print(f"\nHogares con algún gasto extremo: {con_extremo.sum()}")
with pd.option_context('display.float_format', '{:,.0f}'.format):
    display(hogares.loc[con_extremo, ['NOM_VEREDA', 'id_hogar', 'num_personas_hogar'] + dinero]
            .sort_values('vlr_total_gastos', ascending=False))

# 5. Personas con ingresos extremos
print("Honorarios por encima del límite:",
      df_limpio.loc[df_limpio['vlr_ingr_honorarios'] > limites['vlr_ingr_honorarios'],
                    ['NOM_VEREDA', 'edad_calculada', 'vlr_ingr_honorarios']].to_dict('records'))
```

### 6.7 Funciones de apoyo para comparar veredas
Todos los indicadores de la fase 3 se calculan igual: para **cada vereda**, para **La Laguna** (Alta + Baja) y para el **total**. Para no repetir ese código en cada tabla se definen tres funciones:

| Función | Qué hace | Ejemplo |
|---|---|---|
| `por_vereda(df, funcion)` | aplica `funcion` a cada grupo y devuelve una tabla con una fila por grupo | `por_vereda(df_hogares, lambda d: d['num_personas_hogar'].median())` |
| `distribucion(serie, etiquetas)` | porcentaje de cada categoría **sobre quienes respondieron** (sin nulos) | `distribucion(d['sexo_persona'], {1: 'Hombre', 2: 'Mujer'})` |
| `mostrar(tabla)` | presenta la tabla sin decimales innecesarios | `mostrar(tabla_sexo)` |

`distribucion()` descarta los nulos: así cumple la regla de calcular las variables con nulos "por diseño" solo sobre quienes respondieron.

```python
# Grupos de comparación: cada vereda, La Laguna (Alta + Baja) y el total
GRUPOS = {
    'Paso Chico': ['PASOCHICO'],
    'Los Teres': ['LOS TERES'],
    'Laguna Alta': ['LAGUNA ALTA'],
    'Laguna Baja': ['LAGUNA BAJA'],
    'La Laguna': ['LAGUNA ALTA', 'LAGUNA BAJA'],
    'Total': veredas,
}

def por_vereda(df, funcion):
    """Aplica funcion(subconjunto de df) a cada grupo de GRUPOS. Devuelve una fila por grupo."""
    filas = {nombre: funcion(df[df['NOM_VEREDA'].isin(vs)]) for nombre, vs in GRUPOS.items()}
    return pd.DataFrame(filas).T

def distribucion(serie, etiquetas, otros=None):
    """% de cada categoría sobre quienes respondieron, en el orden de `etiquetas`.
    Varios códigos pueden compartir etiqueta (se suman). Los códigos que no estén
    en `etiquetas` se agrupan en `otros` (si se indica)."""
    s = serie.dropna().map(etiquetas)
    orden = list(dict.fromkeys(etiquetas.values()))  # etiquetas sin repetir, en su orden
    if otros:
        s, orden = s.fillna(otros), orden + [otros]
    return (s.value_counts(normalize=True) * 100).reindex(orden, fill_value=0).round(1).add_prefix('% ')

def mostrar(tabla):
    """Presenta la tabla sin decimales innecesarios."""
    return tabla.style.format('{:g}')
```

**Funciones de gráficos.** Todos los indicadores de la fase 3 se grafican con dos tipos de gráfico, para que todas las dimensiones se lean igual:

| Función | Para qué | Cómo se lee |
|---|---|---|
| `barras_apiladas(ax, tabla, columnas, colores, titulo)` | distribuciones que suman 100 % (sexo, grupos de edad, estado civil…) | una barra por grupo; cada color es una categoría; el número dentro es el % (se omite si es menor que 6 %) |
| `barras_valor(ax, serie, titulo)` | un solo indicador por vereda (edad mediana, % de jefas mujeres…) | una barra por vereda; **La Laguna en azul claro** porque es la suma de Alta y Baja; la **línea punteada es el Total** |

**Colores:** categorías sin orden (parentesco, estado civil) con colores distintos (`CATEGORICOS`); categorías ordenadas (grupos de edad, tamaño del hogar) con tonos de un mismo azul, de claro a oscuro (`RAMPA`).

```python
import matplotlib.colors as mcolors

CATEGORICOS = ['#2a78d6', '#eb6834', '#1baf7a', '#eda100', '#e87ba4']  # categorías sin orden
RAMPA = ['#86b6ef', '#5598e7', '#2a78d6', '#1c5cab', '#0d366b']        # categorías ordenadas (claro → oscuro)
COLOR_SEXO = {'Hombre': '#2a78d6', 'Mujer': '#eb6834'}
COLOR_VEREDA, COLOR_LAGUNA = '#2a78d6', '#86b6ef'
TINTA, TINTA_2, SEPARADOR = '#0b0b0b', '#52514e', '#d9d9d6'

def _color_texto(fondo):
    """Texto blanco sobre fondos oscuros y negro sobre fondos claros."""
    r, g, b = mcolors.to_rgb(fondo)
    return 'white' if 0.2126 * r + 0.7152 * g + 0.0722 * b < 0.5 else TINTA

def _estilo(ax, titulo, espacio=22):
    ax.set_title(titulo, loc='left', fontsize=11, color=TINTA, pad=espacio)
    ax.spines[['top', 'right']].set_visible(False)
    ax.tick_params(colors=TINTA_2)
    ax.xaxis.grid(True, alpha=0.25)
    ax.set_axisbelow(True)

def barras_apiladas(ax, tabla, columnas, colores, titulo, min_etiqueta=6):
    """Barras horizontales al 100 %: una barra por grupo y un segmento por categoría."""
    y = np.arange(len(tabla))
    izquierda = np.zeros(len(tabla))
    for col, color in zip(columnas, colores):
        valores = tabla[col].to_numpy(dtype=float)
        ax.barh(y, valores, left=izquierda, color=color, height=0.7,
                edgecolor='white', linewidth=1.5, label=col.replace('% ', ''))
        for yi, v, x0 in zip(y, valores, izquierda):
            if v >= min_etiqueta:
                ax.text(x0 + v / 2, yi, f'{v:.0f}', ha='center', va='center',
                        fontsize=9, color=_color_texto(color))
        izquierda += valores
    ax.axhline(3.5, color=SEPARADOR, lw=1)              # separa las 4 veredas de La Laguna y el Total
    ax.set_yticks(y, tabla.index)
    ax.invert_yaxis()
    ax.set_xlim(0, 100)
    ax.xaxis.set_major_formatter(lambda x, _: f'{x:.0f} %')
    ax.legend(loc='lower left', bbox_to_anchor=(0, 1.0), ncol=len(columnas), frameon=False, fontsize=9)
    _estilo(ax, titulo, espacio=26)                     # más espacio arriba para la leyenda

def barras_valor(ax, serie, titulo, base=0, formato='{:g}'):
    """Una barra por vereda (La Laguna en azul claro). La línea punteada marca el Total.
    `base` es desde dónde empiezan las barras (0, o 100 para el índice de masculinidad)."""
    datos = serie.drop('Total').astype(float)
    total = float(serie['Total'])
    y = np.arange(len(datos))
    colores = [COLOR_LAGUNA if g == 'La Laguna' else COLOR_VEREDA for g in datos.index]
    ax.barh(y, datos - base, left=base, color=colores, height=0.6, zorder=2)
    if base != 0:
        ax.axvline(base, color=TINTA, lw=1, zorder=2)   # línea de base (p. ej. 100 = equilibrio)

    bajo, alto = min(base, datos.min(), total), max(base, datos.max(), total)
    margen = (alto - bajo) * 0.15
    ax.set_xlim(bajo - (margen if bajo < base else 0), alto + margen)

    ax.axvline(total, color=TINTA_2, ls='--', lw=1, zorder=3)
    ax.text(total, -0.6, f'Total: {formato.format(total)}', ha='center', va='bottom', fontsize=8, color=TINTA_2)
    for yi, v in zip(y, datos):                        # fondo blanco para que la línea no tape el número
        ax.text(v + (margen / 4 if v >= base else -margen / 4), yi, formato.format(v),
                va='center', ha='left' if v >= base else 'right', fontsize=9, color=TINTA, zorder=4,
                bbox=dict(facecolor='white', edgecolor='none', pad=1))
    ax.set_yticks(y, datos.index)
    ax.set_ylim(len(datos) - 0.5, -1.1)
    _estilo(ax, titulo)
```

#### 7.1.3 Edad
- **Grupos de edad:** porcentaje de la población de cada vereda en cada grupo.
- **Índice de dependencia:** personas de 0–14 y de 65 o más años por cada 100 personas de 15–64 (edad de trabajar). Más alto = más personas que dependen de quienes están en edad de trabajar.
- **Índice de envejecimiento:** personas de 65 o más años por cada 100 menores de 15. Mayor que 100 = hay más adultos mayores que niños.

```python
BORDES_EDAD = [0, 15, 30, 45, 60, 200]
GRUPOS_EDAD = ['0–14', '15–29', '30–44', '45–59', '60 y más']

def edad(d):
    e = d['edad_calculada']
    ninos, mayores, activos = (e < 15).sum(), (e >= 65).sum(), e.between(15, 64).sum()
    grupos_edad = pd.cut(e, BORDES_EDAD, right=False, labels=GRUPOS_EDAD).value_counts(normalize=True)
    return pd.concat([
        pd.Series({'edad promedio': round(e.mean(), 1), 'edad mediana': e.median()}),
        (grupos_edad.reindex(GRUPOS_EDAD) * 100).round(1).add_prefix('% '),
        pd.Series({'índice de dependencia': round((ninos + mayores) / activos * 100, 1),
                   'índice de envejecimiento': round(mayores / ninos * 100, 1)}),
    ])

tabla_edad = por_vereda(df_personas, edad)
mostrar(tabla_edad)
```

```python
# Gráfico: población por grupo de edad
fig, ax = plt.subplots(figsize=(11, 4))
barras_apiladas(ax, tabla_edad, [f'% {g}' for g in GRUPOS_EDAD], RAMPA, 'Población por grupo de edad (%)')
plt.tight_layout()
plt.show()
```

#### 7.1.4 Parentesco y jefatura del hogar
Parentesco de cada persona con el jefe del hogar. Los parentescos con pocos casos (nietos, padres, hermanos, suegros, otros) se agrupan en "Otro". La **jefatura** y la **presencia de cónyuge** se calculan sobre los **hogares** (un jefe por hogar).

```python
def parentesco(d):
    jefes = d[d['tip_parentesco'] == 1]
    hogar_con_conyuge = d.groupby('id_hogar')['tip_parentesco'].apply(lambda t: (t == 2).any())
    return pd.concat([
        distribucion(d['tip_parentesco'], PARENTESCO, otros='Otro'),
        pd.Series({'% hogares con jefa mujer': round((jefes['sexo_persona'] == 2).mean() * 100, 1),
                   '% hogares con jefe hombre': round((jefes['sexo_persona'] == 1).mean() * 100, 1),
                   'edad mediana del jefe(a)': jefes['edad_calculada'].median(),
                   '% hogares con cónyuge': round(hogar_con_conyuge.mean() * 100, 1)}),
    ])

tabla_parentesco = por_vereda(df_personas, parentesco)
mostrar(tabla_parentesco)
```

#### 7.1.6 Tamaño del hogar
Se calcula sobre `df_hogares` (una fila por hogar) con `num_personas_hogar`. Las **viviendas compartidas** son las que tienen dos o más hogares.

```python
def tamano_hogar(d):
    n = d['num_personas_hogar']
    return pd.Series({
        'hogares': len(d),
        'personas por hogar (promedio)': round(n.mean(), 1),
        'personas por hogar (mediana)': n.median(),
        '% unipersonales (1)': round((n == 1).mean() * 100, 1),
        '% de 2 a 4 personas': round(n.between(2, 4).mean() * 100, 1),
        '% de 5 o más personas': round((n >= 5).mean() * 100, 1),
        '% viviendas con 2 o más hogares': round((d.groupby('num_ficha').size() > 1).mean() * 100, 1),
    })

tabla_hogar = por_vereda(df_hogares, tamano_hogar)
mostrar(tabla_hogar)
```

#### 7.2.5 Discapacidad
Porcentaje de personas con **dificultad permanente** para alguna actividad: ver, oír, hablar, moverse, bañarse o vestirse solo, salir sin ayuda o entender. Una persona puede tener varias.

#### 7.3.2 Actividad principal

La **tasa de búsqueda de trabajo** es el porcentaje de personas que buscan trabajo entre las que trabajan o lo buscan (la fuerza de trabajo). Es parecida a la tasa de desempleo.

```python
def actividad(d):
    a = d['tip_actividad_mes']
    trabaja, busca = (a == 1).sum(), (a == 2).sum()
    return pd.concat([
        pd.Series({'personas de 15 o más': len(d)}),
        distribucion(a, ACTIVIDAD),
        pd.Series({'tasa de búsqueda de trabajo': round(busca / (trabaja + busca) * 100, 1),
                   '% de mujeres que trabajan': round((a[d['sexo_persona'] == 2] == 1).mean() * 100, 1),
                   'semanas buscando (mediana)': d['num_sem_buscando'].median()}),
    ])

tabla_actividad = por_vereda(mayores15, actividad)
mostrar(tabla_actividad)
```

#### 7.3.4 Ingresos del hogar
Se suman, para cada hogar, todos los ingresos mensuales declarados por sus miembros: salario, honorarios o negocio propio, cosecha, pensión, remesas y subsidios.

**Advertencia:** los ingresos están **incompletos**. El 42 % de las personas que trabajan no declara ningún ingreso laboral, y en los hogares que sí declaran ingresos el gasto mediano es 1,4 veces el ingreso. Por eso los ingresos se usan sobre todo para ver **de qué fuentes** viven los hogares, y el **gasto** (7.3.6) es una mejor medida de su capacidad económica.

Para comparar: el salario mínimo de 2023 era **$1.160.000** al mes.

```python
SALARIO_MINIMO = 1_160_000  # 2023
LABORALES = ['vlr_ingr_salario', 'vlr_ingr_honorarios', 'vlr_ingr_cosecha']

# Ingreso de cada hogar = suma de los ingresos de sus miembros
ingreso_hogar = df_personas.groupby('id_hogar')[list(INGRESOS)].sum()
hogares = df_hogares.assign(
    ingreso=df_hogares['id_hogar'].map(ingreso_hogar.sum(axis=1)),
    laboral=df_hogares['id_hogar'].map(ingreso_hogar[LABORALES].sum(axis=1)),
    subsidio=df_hogares['id_hogar'].map(ingreso_hogar[SUBSIDIOS].sum(axis=1)),
)

def ingresos(d):
    return pd.Series({
        '% hogares sin ingresos declarados': round((d['ingreso'] == 0).mean() * 100, 1),
        '% hogares con ingreso laboral': round((d['laboral'] > 0).mean() * 100, 1),
        'ingreso laboral mediano (hogares que lo declaran)': d.loc[d['laboral'] > 0, 'laboral'].median(),
        '% hogares con ingreso de 1 salario mínimo o más': round((d['ingreso'] >= SALARIO_MINIMO).mean() * 100, 1),
        '% del ingreso que viene de subsidios': round(d['subsidio'].sum() / d['ingreso'].sum() * 100, 1),
    })

tabla_ingresos = por_vereda(hogares, ingresos)
print("Hogares con ingreso de cada fuente (%):")
print(((ingreso_hogar > 0).mean() * 100).round(1).rename(INGRESOS).to_string())
mostrar(tabla_ingresos)
```

```python
# Gráfico: hogares sin ingresos declarados
fig, ax = plt.subplots(figsize=(9, 4))
barras_valor(ax, tabla_ingresos['% hogares sin ingresos declarados'], 'Hogares que no declaran ningún ingreso (%)')
plt.tight_layout()
plt.show()
```

#### 7.3.6 Gastos del hogar
Gasto mensual del hogar (`vlr_total_gastos`, la suma de 8 rubros) y el **porcentaje que se va en alimentos**. Cuanto más pobre es un hogar, mayor parte de su gasto dedica a la comida. El **gasto per cápita** es el gasto del hogar dividido entre el número de personas.

```python
def gastos(d):
    total = d['vlr_total_gastos']
    con_gasto = total > 0
    return pd.Series({
        'gasto del hogar (mediana)': total.median(),
        'gasto per cápita (mediana)': round((total / d['num_personas_hogar']).median()),
        '% del gasto en alimentos (mediana)': round((d.loc[con_gasto, 'vlr_gasto_alimento'] / total[con_gasto]).median() * 100, 1),
        '% hogares que pagan arriendo': round((d['vlr_gasto_arriendo'] > 0).mean() * 100, 1),
    })

tabla_gastos = por_vereda(df_hogares, gastos)
mostrar(tabla_gastos)
```

```python
# Gráfico: gasto mensual mediano del hogar (miles de pesos)
fig, ax = plt.subplots(figsize=(9, 4))
barras_valor(ax, tabla_gastos['gasto del hogar (mediana)'] / 1000, 'Gasto mensual mediano del hogar (miles de pesos)',
             formato='{:,.0f}')
plt.tight_layout()
plt.show()
```

#### 7.5.1 Clasificación del Sisbén IV
El Sisbén IV clasifica a cada hogar según su capacidad de generar ingresos y sus condiciones de vida, y define a qué programas sociales puede acceder:
- **A. Pobreza extrema** (subgrupos A1 a A5)
- **B. Pobreza moderada** (B1 a B7)
- **C. Vulnerable** (C1 a C18)
- **D. No pobre ni vulnerable** (D1 a D21)

Se verifica que la columna `Nivel` corresponde al subgrupo y que el grupo es coherente con otros indicadores de pobreza.

```python
GRUPO_SISBEN = {'A': 'A. Pobreza extrema', 'B': 'B. Pobreza moderada', 'C': 'C. Vulnerable',
                'D': 'D. No pobre ni vulnerable'}

# Verificación
h = df_hogares
print("1. Subgrupo máximo (Nivel) por grupo:", h.groupby('Grupo')['Nivel'].max().to_dict(),
      "| en el Sisbén IV: A hasta 5, B hasta 7, C hasta 18, D hasta 21")
print("2. % con piso de tierra por grupo:",
      h.groupby('Grupo')['tip_mat_pisos'].apply(lambda s: round((s == 5).mean() * 100, 1)).to_dict())
print("3. Gasto mediano del hogar por grupo:", h.groupby('Grupo')['vlr_total_gastos'].median().to_dict())

def sisben(d):
    g = d['Grupo']
    return pd.concat([
        pd.Series({'hogares': len(d)}),
        distribucion(g, GRUPO_SISBEN),
        pd.Series({'% en pobreza (A + B)': round(g.isin(['A', 'B']).mean() * 100, 1)}),
    ])

tabla_sisben = por_vereda(df_hogares, sisben)
mostrar(tabla_sisben)
```

#### 7.5.2 Eventos y desastres
Porcentaje de hogares que reportan haber sufrido cada tipo de evento: inundación, terremoto o temblor, incendio, vendaval y hundimiento de terreno. Cada evento tiene pocos casos, así que solo se compara el total de hogares afectados por alguno.

```python
EVENTOS = {'ind_evento_terremoto': 'terremoto o temblor', 'ind_evento_vendaval': 'vendaval',
           'ind_evento_hundimiento': 'hundimiento', 'ind_evento_incendio': 'incendio',
           'ind_evento_inundacion': 'inundación'}

def eventos(d):
    sufrio = d[list(EVENTOS)] == 1
    return pd.concat([
        pd.Series({'hogares afectados': sufrio.any(axis=1).sum(),
                   '% afectado por algún evento': round(sufrio.any(axis=1).mean() * 100, 1)}),
        (sufrio.mean() * 100).round(1).rename(lambda c: f'% {EVENTOS[c]}'),
    ])

tabla_eventos = por_vereda(df_hogares, eventos)
mostrar(tabla_eventos)
```

### 8.1 Tabla resumen
28 indicadores agrupados por dimensión. En cada fila, entre las 4 veredas, el valor **más alto** aparece en naranja y el **más bajo** en azul. El color no indica si es bueno o malo: depende del indicador.

```python
VEREDAS_4 = ['Paso Chico', 'Los Teres', 'Laguna Alta', 'Laguna Baja']

# Indicadores de la síntesis, agrupados por dimensión
INDICADORES = {
    'Tamaño': ['personas', 'hogares'],
    'D1. Población': ['edad mediana', '% 0–14', '% 60 y más', 'índice de dependencia', 'índice de envejecimiento',
                      '% hogares con jefa mujer', '% hogares con cónyuge', 'personas por hogar (promedio)',
                      '% unipersonales (1)'],
    'D2. Educación y salud': ['% no sabe leer ni escribir (15+)', '% sin nivel educativo (25+)',
                              '% con media o más (25+)', '% régimen subsidiado', '% con alguna discapacidad'],
    'D3. Economía del hogar': ['% trabajadores que cotizan a pensión', '% hogares sin ingresos declarados',
                               '% hogares con algún subsidio', 'gasto mensual del hogar (mediana)',
                               '% del gasto en alimentos'],
    'D4. Vivienda y servicios': ['% vivienda propia', '% piso de tierra', '% sin servicio sanitario',
                                 '% cocina con leña', '% hacinamiento crítico', '% con nevera'],
    'D5. Vulnerabilidad': ['% pobreza extrema (Sisbén A)', '% en pobreza (Sisbén A + B)',
                           '% afectado por algún evento'],
}

claves = pd.concat([clave_D1, clave_D2, clave_D3, clave_D4, clave_D5], axis=1)
filas = [(dimension, indicador) for dimension, lista in INDICADORES.items() for indicador in lista]
resumen = claves[[indicador for _, indicador in filas]].T
resumen.index = pd.MultiIndex.from_tuples(filas, names=['dimensión', 'indicador'])

def resaltar(fila):
    """Naranja para el valor más alto y azul para el más bajo entre las 4 veredas."""
    if fila.name[0] == 'Tamaño':
        return [''] * len(fila)
    valores = fila[VEREDAS_4]
    estilo = {valores.idxmax(): 'background-color: #fbd9c9', valores.idxmin(): 'background-color: #cde2fb'}
    return [estilo.get(columna, '') for columna in fila.index]

resumen.style.format('{:,.6g}').apply(resaltar, axis=1)
```

### 8.2 Mapa de diferencias
Cada celda muestra el **valor** del indicador en la vereda. El **color** indica cuánto se aleja ese valor del **total** de las 4 veredas: **naranja** si está por encima, **azul** si está por debajo y gris claro si es parecido. Así se ve de un vistazo qué distingue a cada vereda.

```python
from matplotlib.colors import LinearSegmentedColormap, Normalize

dimensiones = [d for d in INDICADORES if d != 'Tamaño']
indicadores = [i for d in dimensiones for i in INDICADORES[d]]
valores = claves.loc[VEREDAS_4, indicadores].T          # filas = indicadores, columnas = veredas
total = claves.loc['Total', indicadores]
diferencia = (valores.sub(total, axis=0).div(total, axis=0) * 100).clip(-60, 60)  # % sobre el total

def texto(v):
    return f'{v / 1000:.0f} mil' if v >= 1000 else f'{v:g}'

colores = LinearSegmentedColormap.from_list('diferencia', ['#2a78d6', '#f0efec', '#eb6834'])
escala = Normalize(vmin=-60, vmax=60)

fig, ax = plt.subplots(figsize=(10, 12))
imagen = ax.imshow(diferencia.values, cmap=colores, norm=escala, aspect='auto')
for i in range(len(indicadores)):
    for j in range(len(VEREDAS_4)):
        fondo = colores(escala(diferencia.iat[i, j]))
        ax.text(j, i, texto(valores.iat[i, j]), ha='center', va='center', fontsize=10, color=_color_texto(fondo))

ax.set_xticks(range(len(VEREDAS_4)), VEREDAS_4)
ax.xaxis.tick_top()
ax.set_yticks(range(len(indicadores)), [f'{nombre}  (total: {texto(total[nombre])})' for nombre in indicadores])

# Líneas blancas que separan las dimensiones y nombre de cada dimensión a la derecha
inicio = 0
for d in dimensiones:
    fin = inicio + len(INDICADORES[d])
    if inicio:
        ax.axhline(inicio - 0.5, color='white', lw=4)
    ax.text(len(VEREDAS_4) - 0.4, (inicio + fin - 1) / 2, d, va='center', ha='left', fontsize=10, color=TINTA)
    inicio = fin

ax.tick_params(colors=TINTA_2, length=0)
for borde in ax.spines.values():
    borde.set_visible(False)

# Barra de color horizontal justo debajo del mapa
cax = ax.inset_axes([0.2, -0.04, 0.6, 0.012])
barra = fig.colorbar(imagen, cax=cax, orientation='horizontal')
barra.set_label('% por encima (+) o por debajo (−) del total', color=TINTA_2)
ax.set_title('Qué distingue a cada vereda', loc='left', fontsize=14, color=TINTA, pad=36)
plt.tight_layout()
plt.show()
```
