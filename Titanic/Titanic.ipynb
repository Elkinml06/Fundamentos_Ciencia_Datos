```python
#Librerias necesarias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Cargar dataset
ruta='/content/drive/MyDrive/U/Fundamentos de Ciencia de Datos/Titanic/titanic.csv'
df = pd.read_csv(ruta)
df.head()
```

```python
#Ver si cuenta con datos nulos
df.isnull().sum()
```

```python
print(df["Survived"].unique())
print(df["Sex"].unique())
print(df["Pclass"].unique())
print(df["Embarked"].unique())
```

```python
df_clean=df.copy()
```

```python
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
```

```python
df_clean["Embarked"] = df_clean["Embarked"].fillna(
    df_clean["Embarked"].mode()[0]
)
```

```python
df_clean["Cabin"] = df_clean["Cabin"].fillna("None")
```

```python
df_clean.isnull().sum()
```

```python
def limpiar_ticket_simple(ticket):
  ticket = str(ticket).strip()

  # Si tiene un espacio, nos quedamos únicamente con la última parte (el número)
  if ' ' in ticket:
    return ticket.rsplit(' ', 1)[-1]

  # Si es solo número o no tiene espacios, lo dejamos tal cual
  return ticket


# Aplicar la función directamente sobre la columna Ticket
df_clean['Ticket'] = df_clean['Ticket'].apply(limpiar_ticket_simple)
```

```python
df_clean.head()
```

```python
# 1. Filtramos mujeres de primera clase (Rose)
rose_subset = df_clean[
    (df_clean['Sex'] == 'female') &
    (df_clean['Pclass'] == 1)
]

# 2. Filtramos hombres de tercera clase (Jack)
jack_subset = df_clean[
    (df_clean['Sex'] == 'male') &
    (df_clean['Pclass'] == 3)
]

# 3. Calculamos las tasas de supervivencia
tasa_rose = rose_subset['Survived'].mean() * 100
tasa_jack = jack_subset['Survived'].mean() * 100

# 4. Mostramos los resultados
print("=== COMPARACIÓN DE SUPERVIVENCIA ===\n")

print(f"Rose (mujeres de primera clase): {tasa_rose:.2f}%")
print(f"Jack (hombres de tercera clase): {tasa_jack:.2f}%")

print(f"\nPasajeras del grupo de Rose: {len(rose_subset)}")
print(f"Pasajeros del grupo de Jack: {len(jack_subset)}")
```

```python
print("=== CONCLUSIÓN ===\n")
print(f"1. El grupo de Rose (primera clase, mujeres) representa la seguridad de la puerta, con una supervivencia del {tasa_rose:.2f}%.")
print("2. Si Jack se sube a la puerta con Rose, adopta este mismo entorno seguro de supervivencia.")
print(f"3. Por lo tanto, la probabilidad de Jack cambia de un bajo 13.54% a un casi seguro {tasa_rose:.2f}%.")
print("\nResultado final: SÍ, SE HUBIERA SALVADO.")
```
